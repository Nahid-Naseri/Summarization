## This notebook outlines the concepts behind Text Summarization

## Summarization
- concept of capturing very important gist of a long piece of text

### Types of Summarization
- 1. **Extractive Summarization**
    - Select sentences from the corpus that best represent the text
    - Arrange them to form a summary
- 2. **Abstractive Summarization**
    - Captures the very important sentences from the text
    - Paraphrases them to form a summary

```python
# Fetch the webpage content
response = requests.get(url)
html_content = response.content

# Create an HtmlParser object
parser = HtmlParser.from_string(html_content, url, Tokenizer("english"))

# Extract the text from the parser - iterating through sentences.
text = ""
# Access sentences using document.sentences
for sentence in parser.document.sentences: # Use parser.document.sentences to access the iterable sentences
    text += str(sentence)
```

```python
def get_page(url):
    """
    Fetches the webpage content and parses it using BeautifulSoup.

    Args:
        url (str): The URL of the webpage to fetch.

    Returns:
        bs4.BeautifulSoup: The parsed webpage content.
    """
    response = requests.get(url)
    response.raise_for_status()  # Raise an exception for bad responses
    soup = BeautifulSoup(response.content, "html.parser")
    return soup
```

```python
def collect_text(soup):
    """
    Extracts the text content from the webpage.

    Args:
        soup (bs4.BeautifulSoup): The parsed webpage content.

    Returns:
        str: The extracted text content.
    """
    text = ""
    for paragraph in soup.find_all("p"):
        text += paragraph.get_text()
    return text
```

```python
# Install necessary libraries
!pip install --upgrade numpy
!pip install --force-reinstall summa
!pip install nltk
!pip install beautifulsoup4
!pip install sumy
!pip install rouge-score
import requests
from bs4 import BeautifulSoup
from sumy.parsers.plaintext import PlaintextParser
from sumy.nlp.tokenizers import Tokenizer
from sumy.summarizers.text_rank import TextRankSummarizer
from sumy.summarizers.lex_rank import LexRankSummarizer
from sumy.summarizers.luhn import LuhnSummarizer
from sumy.summarizers.lsa import LsaSummarizer
from summa.summarizer import summarize  # Using summa
import nltk

# Download NLTK data
nltk.download('punkt')

# Function to fetch and extract article text
def fetch_article_text(url):
    response = requests.get(url)
    response.raise_for_status()
    soup = BeautifulSoup(response.content, "html.parser")
    paragraphs = soup.find_all("p")
    article_text = "\n".join([p.get_text() for p in paragraphs])
    return article_text

# URL of the Medium article
url = "https://medium.com/@subashgandyer/papa-what-is-a-neural-network-c5e5cc427c7"
article_text = fetch_article_text(url)

# Summarization using Sumy
parser = PlaintextParser.from_string(article_text, Tokenizer("english"))

def summarize_with_sumy(summarizer_class, parser, sentence_count=7):
    summarizer = summarizer_class()
    summary = summarizer(parser.document, sentences_count=sentence_count)
    return " ".join(str(sentence) for sentence in summary)

# Generate summaries using different methods
summaries = {
    "TextRank": summarize_with_sumy(TextRankSummarizer, parser),
    "LexRank": summarize_with_sumy(LexRankSummarizer, parser),
    "Luhn": summarize_with_sumy(LuhnSummarizer, parser),
    "LSA": summarize_with_sumy(LsaSummarizer, parser),
    "Summa": summarize(article_text, ratio=0.1),  # Summarization using summa
}

# Print summaries
for method, summary in summaries.items():
    print(f"\n{method} Summary:\n", summary)
```

```python
# Install necessary packages
!pip install rouge-score

from rouge_score import rouge_scorer

# Function to calculate ROUGE scores
def calculate_rouge(reference, summary):
    scorer = rouge_scorer.RougeScorer(['rouge1', 'rouge2', 'rougeL'], use_stemmer=True)
    scores = scorer.score(reference, summary)
    return scores

# Original article text (reference)
reference = article_text

# Summarized texts (generated summaries)
summaries = {
    "TextRank": summarize_with_sumy(TextRankSummarizer, parser),
    "LexRank": summarize_with_sumy(LexRankSummarizer, parser),
    "Luhn": summarize_with_sumy(LuhnSummarizer, parser),
    "LSA": summarize_with_sumy(LsaSummarizer, parser),
    "Summa": summarize(article_text, ratio=0.1),
}

# Calculate ROUGE scores for all summaries and store them
rouge_scores = {}

for method, summary in summaries.items():
    scores = calculate_rouge(reference, summary)
    rouge_scores[method] = scores

# Open file to save results
with open("Nahid_Task4_summery.txt", "w") as f:
    # Add summaries of all methods
    for method, summary in summaries.items():
        f.write(f"\n{method} Summary:\n")
        f.write(summary + "\n\n")

    # Add ROUGE scores
    for method, scores in rouge_scores.items():
        f.write(f"{method} ROUGE Scores:\n")
        f.write(f"ROUGE-1: {scores['rouge1']}\n")
        f.write(f"ROUGE-2: {scores['rouge2']}\n")
        f.write(f"ROUGE-L: {scores['rougeL']}\n\n")

    # Optionally, choose the best method based on ROUGE scores (ROUGE-1 F-measure)
    best_method = max(rouge_scores, key=lambda x: rouge_scores[x]['rouge1'].fmeasure)
    best_summary = summaries[best_method]
    justification = f"{best_method} provided the best ROUGE-1 F-measure."

    # Save the best summary and justification to the file
    f.write(f"Best Summary Method: {best_method}\n\n")
    f.write(best_summary + "\n\n")
    f.write("Justification: " + justification)

    # Add observation at the end of the file
    observation = """
    Observation:
    Based on the ROUGE scores, **Summa** provided the best summary with the highest ROUGE-1 F-measure.
    This indicates that Summa was able to preserve the most important words from the original article.
    The higher the ROUGE-1 F-measure, the better the summary reflects the original content in terms of word coverage.
    Other methods like TextRank and Luhn also performed well, but Summa stood out in overall accuracy and relevance.
    Conclusion:
    Summa provides the best balance between precision and recall, preserving more key words from the original text (as shown by its highest ROUGE-1 F-measure). This makes it the most effective summarization method among the ones tested in this evaluation.
    """

    f.write(observation)


print("Summary and ROUGE scores have been saved to 'Nahid_Task4_summery.txt'.")
```
